```python
from sklearn.model_selection import train_test_split
# Using y_arr instead of y because y was overwritten with its shape tuple
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.23, random_state=42)
```

```python
model = keras.Sequential([
    layers.Input(shape=(251,1)),
    layers.LSTM(512,return_sequences=True),
    layers.LSTM(256),
    layers.Dense(8646,activation="softmax")
])
```

```python
model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
from tensorflow.keras.callbacks import EarlyStopping

early_stop = EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

history = model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=30,
    batch_size=32,
    #callbacks=[early_stop]
)
```

```python
def preprocess_seed(seed_text):
    seed_text = seed_text.lower()
    seed_text = ''.join(re.findall(r'[a-z\s]', seed_text)).strip()
    return ' '.join([lema.lemmatize(w) for w in seed_text.split()])


def predict_next_word(seed_text, top_k=3):
    """Return the top_k most likely next words with their probabilities."""
    seed = preprocess_seed(seed_text)
    token_list = tok.texts_to_sequences([seed])[0]
    if len(token_list) == 0:
        return []
    token_list = pad_sequences([token_list], maxlen=251, padding='post')
    probs = model.predict(token_list, verbose=0)[0]
    top_idx = np.argsort(probs)[::-1][:top_k + 1]
    results = []
    for idx in top_idx:
        word = tok.index_word.get(int(idx))
        if word is not None:          # index 0 is padding and has no word
            results.append((word, float(probs[idx])))
    return results[:top_k]


def generate_text(seed_text, n_words=5):
    """Repeatedly predict the next word and append it to the sentence."""
    current = seed_text
    for _ in range(n_words):
        preds = predict_next_word(current, top_k=1)
        if not preds:
            break
        current += ' ' + preds[0][0]
    return current
```

```python
import random
random.seed(7)

words = lem_text.split()
seed_len = 4
n_tests = 8
hits = 0

for _ in range(n_tests):
    start = random.randint(0, len(words) - seed_len - 2)
    seed = ' '.join(words[start:start + seed_len])
    actual = words[start + seed_len]
    top3 = predict_next_word(seed, top_k=3)
    pred_words = [w for w, _ in top3]
    hits += actual in pred_words
    print(f"Input      : {seed}")
    print(f"Actual next: {actual}")
    print(f"Top 3      : {', '.join(f'{w} ({p:.2f})' for w, p in top3)}")
    print(f"Generated  : {generate_text(seed, n_words=5)}")
    print('-' * 60)

print(f"Actual word found in top 3 predictions: {hits}/{n_tests}")
```
